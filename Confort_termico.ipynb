## Teoría del Confort Térmico en Ganado Vacuno

El confort térmico en el ganado vacuno se refiere a un estado en el que los animales no experimentan estrés por calor ni por frío, lo que les permite mantener su temperatura corporal central dentro de un rango óptimo con un mínimo esfuerzo metabólico. Este estado es crucial para la salud, el bienestar y la productividad del ganado (producción de leche, carne, reproducción).

### Factores que Afectan el Confort Térmico:

1.  **Temperatura del Aire (Tbs - Temperatura de Bulbo Seco):** Es el factor más obvio. Temperaturas extremas, tanto altas como bajas, pueden causar estrés.
2.  **Humedad Relativa (Tbh - Temperatura de Bulbo Húmedo):** Una alta humedad reduce la capacidad de los animales para disipar calor a través de la evaporación (sudoración y jadeo), lo que es especialmente crítico en ambientes cálidos.
3.  **Radiación Solar:** La exposición directa al sol aumenta significativamente la carga de calor en los animales.
4.  **Velocidad del Viento:** El viento puede ayudar a disipar el calor del cuerpo en ambientes cálidos (enfriamiento por convección), pero puede aumentar la pérdida de calor en ambientes fríos.
5.  **Precipitación:** La lluvia o la nieve pueden alterar la capa aislante del pelaje, afectando la termorregulación.

### Impacto del Estrés Térmico:

El estrés térmico, particularmente el estrés por calor, tiene efectos negativos significativos en el ganado vacuno, incluyendo:

*   **Reducción en la Producción de Leche:** Disminución en el volumen y calidad de la leche.
*   **Menor Ganancia de Peso:** Afecta la eficiencia de conversión alimenticia en ganado de engorde.
*   **Problemas Reproductivos:** Reducción de la fertilidad, aumento de abortos y ciclos estrales irregulares.
*   **Mayor Susceptibilidad a Enfermedades:** Debilitamiento del sistema inmunológico.
*   **Cambios de Comportamiento:** Aumento de la respiración (jadeo), búsqueda de sombra, menor consumo de alimento.

### Índices de Confort Térmico:

Para cuantificar y predecir el nivel de estrés térmico, se utilizan diversos índices, como el **Índice de Temperatura y Humedad (ITH)**. Estos índices combinan diferentes variables meteorológicas en una sola medida que correlaciona con la respuesta fisiológica del animal al ambiente. El ITH es uno de los más comunes y se utiliza para establecer umbrales que indican niveles de confort, alerta, peligro y situación crítica para el ganado.

# **Métodos y cálculos**

+ Índice de temperatura y humedad (producción intensiva)

$$ITH = 0.72 * (T_{bs} + T_{bs}) + 40.6$$

Donde:

1. Tbs: Temperatura bulbo seco (°C)
2. Tbh: Temperatura de bulbo húmedo (°C)

Para la correlación de los datos de la especie bovina en la producción de leche se recomienda
utilizar los límites propuestos por National Weather Service:

1. Sí $ITH <= 70$ ambiente no estresante
2. Sí $ITH > 70 \space and \space $ITH <= 78$ situación de alerta por estrés térmico

3. Sí $ITH > 78 \space and \space <= 83$ peligro por estrés térmico


4. Sí $ITH > 83$ situación crítica

````python
# ENTRADA: pedir tbs (Temperatura de Bulbo seco) y tbh (Temperatura de bulbo humedo)
tbs = float(input("Ingrese la temperatura de bulbo seco, Grados centigrados: "))
tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados centigrados: "))

# PROCESO: Calculo de ith (Indice de Temperatura y humedad)
ith = 0.72 * (tbs + tbh) + 40.6

# SALIDA: Imprimir resultado, de la evaluación del ith y el confort térmico
if ith <=70:
  print("El ambiente no estresante")
elif ith > 70 and ith <= 78:
  print("situacion de alerta por estrés térmico")
elif ith > 78 and ith <= 83:
  print("peligro por estrés térmico")
else:
  print("situacion critica")
````
